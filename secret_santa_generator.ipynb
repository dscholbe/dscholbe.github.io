```python
def santa_race():
    import random
    import json
    import pandas as pd

    # Configuration
    santas = [f"santa{i}" for i in range(1, 11)]
    goal = 3
    min_step = 0.02
    max_step = 0.25

    # Storage for movements
    movements = []
    positions = {santa: 0.0 for santa in santas}

    # Simulate the race
    while max(positions.values()) < goal:
        round_moves = []
        for santa in santas:
            step = random.uniform(min_step, max_step)
            # Prevent going past the goal
            if positions[santa] + step > goal:
                step = goal - positions[santa]
            positions[santa] += step
            round_moves.append({"id": santa, "move": step})
        movements.append(round_moves)

    # Save race movements JSON
    race_data = {"movements": movements}
    with open("santa_race.json", "w") as f:
        json.dump(race_data, f, indent=2)

    # Create race movements DataFrame (each row = round, each column = santa's move)
    df_race = pd.DataFrame([
        {m["id"]: m["move"] for m in round_moves} for round_moves in movements
    ])

    # Calculate final standings
    final_standings = sorted(
        [{"id": santa, "total": pos} for santa, pos in positions.items()],
        key=lambda x: x["total"], reverse=True
    )

    # Add ranks
    for rank, entry in enumerate(final_standings, start=1):
        entry["rank"] = rank

    # Save final standings JSON
    final_data = {"final_standings": final_standings}
    with open("santa_final_standings.json", "w") as f:
        json.dump(final_data, f, indent=2)

    # Create final standings DataFrame
    df_final = pd.DataFrame(final_standings)
    return df_race, df_final

df_race, df_final = santa_race()
santa_mappings_dict = {
    'santa1': 'an', 
    'santa2': 'dad', 
    'santa3': 'dawson',
    'santa4': 'donny',
    'santa5': 'katherine',
    'santa6': 'linda',
    'santa7': 'maggie',
    'santa8': 'mom',
    'santa9': 'owain',
    'santa10': 'xuyen'
}

df_race = df_race.rename(columns=santa_mappings_dict)
df_final['id'] = df_final['id'].map(santa_mappings_dict)
```
